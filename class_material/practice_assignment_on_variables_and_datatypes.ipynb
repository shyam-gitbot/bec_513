3. A number is divisible by 3 only if the sum of the digits of that number is divisible by 3. Write a code to get sum of the digits of a 4-digit number and then check if the sum is divisble by 3.

```python
n = (input("Enter a 4-digit number: "))
while len(n)!=4:
    print("Error!!!")
    n = input("Enter the number again : ") 
print("The number you gave us is :",n)
summ=0
for i in n:
    summ +=int(i)
    # print('check')
if summ%3==0:
    print("it is divisible")
else:
    print('kya yaar')
```
