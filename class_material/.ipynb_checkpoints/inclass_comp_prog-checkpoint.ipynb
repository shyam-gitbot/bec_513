```python
n = int(input("enter the number to check if it is prime or not :"))
a = False
for i in range(2,int((n**0.5))+1):
    if n%i==0:
        print("its not a prime")
        a = True
        break

# print("its a prime")
if not a and n>1:
    print("its a prime")
elif not a: print("number out of range")




def isPrime(n):
    for i in range(2,int(n**0.5)+1):
        if n%i==0: return False
    return True
print(isPrime(n))
```
