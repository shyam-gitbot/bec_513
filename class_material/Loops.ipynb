## Loops

We have learned lists and dictionaries in the last two sessions. Now it is the time to put them into use.

When you have list/dictionary or for that matter any iterable object, it is natural that you want to walk through them and get what you want. Exmaples could be, i) checking for existence of an element, ii) extracting a subset of elements from the list based on conditions 

```python
s = 0
while True:
    a = input()
    if a !='end': 
        s = s + float(a)
    else:
        break
print (s)
```

```python
from collections import defaultdict
counter = 0
counter_dictionary = defaultdict(lambda : 0)
dna_molecule_file_pointer = open("dna_molecule_lengths.tsv", "r")
output_fp = open("frequency.tsv", "w")
for line in dna_molecule_file_pointer:
    counter_dictionary[line.strip()] +=1
    counter +=1 

dna_molecule_file_pointer.close()

for key, value in counter_dictionary.items():
    output_fp.write(key + "\t" + str(value) + "\t" + str(round(value/counter, 5)) + "\n")

output_fp.close()
```
